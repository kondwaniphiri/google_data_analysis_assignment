```python
# Drop rows with NaNs in 'review' column
reviews_df.dropna(subset=['reviewText'], inplace=True)
```

```python
from spellchecker import SpellChecker
import pandas as pd
from multiprocessing import Pool, cpu_count

# Initialize spell checker
spell = SpellChecker()

# Function to correct spelling
def correct_spelling(text):
    if pd.isnull(text):  # Handle missing values
        return text
    words = text.split()
    corrected_words = [spell.correction(word) if spell.correction(word) else word for word in words]
    return " ".join(corrected_words)

# Parallel processing
def parallel_spell_check(texts):
    with Pool(cpu_count()) as pool:  # Use max available CPU cores
        corrected_texts = pool.map(correct_spelling, texts)
    return corrected_texts

# Apply optimized spell check
#reviews_df['reviewText'] = parallel_spell_check(reviews_df['reviewText'].tolist())
```

```python
#remove contractions
```

```python
import re

# Dictionary of common contractions
contraction_map = {
    "I'm": "I am", "you're": "you are", "he's": "he is", "she's": "she is", "it's": "it is", "we're": "we are",
    "they're": "they are", "I've": "I have", "you've": "you have", "we've": "we have", "they've": "they have",
    "can't": "cannot", "won't": "will not", "don't": "do not", "didn't": "did not", "shouldn't": "should not",
    "wouldn't": "would not", "couldn't": "could not", "isn't": "is not", "aren't": "are not", "wasn't": "was not",
    "weren't": "were not", "hasn't": "has not", "haven't": "have not", "hadn't": "had not"
}

# Function to expand contractions using regex
def expand_contractions(text):
    if pd.isnull(text):  # Handle missing values
        return text
    for contraction, full_form in contraction_map.items():
        text = re.sub(r'\b' + contraction + r'\b', full_form, text)
    return text

# Apply the function
reviews_df['reviewText'] = reviews_df['reviewText'].apply(expand_contractions)

# Print results
print(reviews_df[['reviewText']])
```

```python
#tockenize review column
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Function to tokenize words
def tokenize_words(text):
    if pd.isnull(text):  # Handle missing values
        return text
    return word_tokenize(text)

# Function to tokenize sentences
def tokenize_sentences(text):
    if pd.isnull(text):  # Handle missing values
        return text
    return sent_tokenize(text)

# Apply tokenization to DataFrame
reviews_df['WordTokens'] = reviews_df['reviewText'].apply(tokenize_words)
reviews_df['SentenceTokens'] = reviews_df['reviewText'].apply(tokenize_sentences)

# Print results
print(reviews_df[['WordTokens', 'SentenceTokens']])
```

```python
#Top 20 3-word phrases from 5-star reviews
```

```python
from collections import Counter
five_star_reviews = reviews_df[reviews_df['rating'] == 5.0]['reviewText']

threeWord = []
for review in five_star_reviews:
    tokens = review.split()
    threeWord += [' '.join(tokens[i:i+3]) for i in range(len(tokens)-2)]

top_threeWords = Counter(threeWord).most_common(20)


top_threeWords
```

```python
#Sentiment Classification of Reviews:
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()

def classify_sentiment(text):
    score = sia.polarity_scores(text)
    if score['compound'] >= 0.05:
        return 'Positive'
    elif score['compound'] <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

reviews_df['sentiment'] = reviews_df['reviewText'].apply(classify_sentiment)
sentiment_counts = reviews_df['sentiment'].value_counts()
sentiment_counts
```
